```python
# verifying the stages in order
stages = [
    "visited_site",
    "signup_started",
    "details_filled",
    "email_verified",
    "purchase_completed"
]
df_stages = df[df["step"].isin(stages)]

df_stages
```

```python
users = []

for stage in stages:
    count = df[df["step"] == stage]["user_id"].nunique()
    users.append(count)
print("Users at each stage")
for i in range(len(stages)):
    print(stages[i], ":", users[i])
```

```python
#Conversion rate logic
conversion = [100]
for i in range(1, len(users)):
    rate = (users[i] / users[i-1]) * 100
    conversion.append(rate)
print("\nConversion Rate")
for i in range(len(stages)):
    print(stages[i], ":", round(conversion[i],2), "%")

# Biggest Drop-off
largest = 0
drop_stage = ""
for i in range(1, len(users)):
    lost = users[i-1] - users[i]

    if lost > largest:
        largest = lost
        drop_stage = stages[i-1] + " → " + stages[i]

print("\nBiggest Drop-off")
print(drop_stage)
print("Users Lost:", largest)
```

```python
stages = [
    "visited_site",
    "signup_started",
    "details_filled",
    "email_verified",
    "purchase_completed"
]

print("Average Time Between Consecutive Stages\n")

for i in range(len(stages)-1):


    stage1 = df[df["step"] == stages[i]][["user_id","timestamp"]]

    stage2 = df[df["step"] == stages[i+1]][["user_id","timestamp"]]

    merged = pd.merge(stage1, stage2, on="user_id")

    merged["time_taken"] = merged["timestamp_y"] - merged["timestamp_x"]

    print(stages[i], "→", stages[i+1])
    print("Average Time:", merged["time_taken"].mean())
    print()
```

```python
# Find the biggest drop-off

max_drop = 0
drop_stage = ""

for i in range(1, len(users)):

    drop = users[i-1] - users[i]

    if drop > max_drop:
        max_drop = drop
        drop_stage = stages[i-1] + " → " + stages[i]

print("Alert : ")
print("Biggest Drop-off Stage :", drop_stage)
print("Users Lost :", max_drop)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(stages, users)

plt.title("User Funnel")
plt.xlabel("Stages")
plt.ylabel("Users")

for i in range(len(users)):
    plt.text(i, users[i], users[i], ha="center")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(stages, conversion, marker='o')

plt.title("Conversion Rate")
plt.xlabel("Stages")
plt.ylabel("Conversion Rate (%)")

for i in range(len(conversion)):
    plt.text(i, conversion[i], str(round(conversion[i],1)) + "%", ha='center')

plt.grid(True)

plt.show()
```
